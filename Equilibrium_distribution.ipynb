```python
import numpy as np
from scipy.special import gamma
import matplotlib.pyplot as plt
```

```python
def p_eq(r, nu, lambda_r, epsilon):
    a = np.power(-epsilon**2/nu, 1/2-lambda_r/epsilon**2)
    b = 1/gamma(lambda_r/epsilon**2 - 1/2)
    c = np.power(r, (2*lambda_r-epsilon**2)/epsilon**2)
    d = np.exp(nu*r**2/epsilon**2)

    return 2 * a * b * c * d
```

```python
# plot

nu = -1
lambda_r = 5
epsilon = 0.2

r_max = np.sqrt((lambda_r - epsilon**2)/(-nu))

fig, ax = plt.subplots(1, 1, figsize = (5, 4))
fig.subplots_adjust(left = 0.1, right = 0.98, bottom = 0.1, top = 0.85, wspace = 0.25)
fig.suptitle(rf"$\nu$ = {nu}, $\lambda_r$ = {lambda_r}, $\epsilon$ = {epsilon}")

r = np.arange(0.1, 6, 0.01)

p_eq_r = p_eq(r, nu, lambda_r, epsilon)

ax.plot(r, p_eq_r)

ax.vlines(r_max, 0, p_eq(r_max, nu, lambda_r, epsilon), ls = "--", color = "green", label = r"$r_{max}$")

ax.set_xlabel("r")
ax.set_ylabel(r"$p_{eq}(r)$")
ax.legend()

plt.savefig(f"../Figures/Stochasticity/Equilibrium_distribution.png")
```
